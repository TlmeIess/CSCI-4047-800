### 1. Build a DataFrame from a dict
Make a DataFrame `people` with columns name ('Sam','Dana') and age (20, 22). Show it.

```python
# hint: pd.DataFrame({"name": [...], "age": [...]})
import pandas as pd

df = people = pd.DataFrame({
    "name": ["Sam", "Dana"],
    "age": [25, 30]
})
print(people)
```
