### 1. Variables and types
Create a variable `price` set to 42.50 and print its type.

### 2. Arithmetic
A meal costs 42.50. Print what a 20% tip would be, then the total with tip.

```python
# hint: 20% of price is price * 0.20
total = price * 0.20 + price
print(total)
```

### 4. Comparison

### 5. if / elif / else
Given a variable `score = 73`, print "A" if 90+, "B" if 80-89, "C" if 70-79, else "Below C".

```python
score = 73
# hint: if / elif / elif / else, using >=
if score > 90:
    print("A")
elif score > 80:
    print("B")
elif score > 70:
    print("C")
```

### 6. for loop
Given `scores = [88, 92, 70, 61, 95]`, print each score on its own line.

```python
scores = [88, 92, 70, 61, 95]
# hint: for s in scores:
for s in scores: 
    print(f"score: {s}")
```

### 7. Function (the challenge)
Write a function `total_with_tip(bill)` that returns the bill plus an 18% tip. Test it on 42.50.

```python
# hint: def total_with_tip(bill): return ...
def total_with_tip(bill, percent_tip=0.18):
    return bill * (1 + percent_tip)
print(total_with_tip(42.50))
```
