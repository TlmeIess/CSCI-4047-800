### 7. Comprehension (the challenge)
Given `bills = [20, 35, 50]`, build a list `totals` where each is the bill plus a 20% tip. Print it.

```python
bills = [20, 35, 50]
# hint: [b + b*0.20 for b in bills]
tips = []
for b in bills:
    tip = b * 0.20
    tips.append(tip)
print("tip amount", tips)

total = []
for b in bills:
    total_bill = b * 1.20
    total.append(total_bill)
print("total bill cost:", total)

print("__________________________________")
# This is how it the code is shown in answer sheet but both ways work
bills = [20, 35, 50]
totals = [b + b*0.20 for b in bills]
print(totals)
```
