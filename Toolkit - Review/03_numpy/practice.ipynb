### 1. Make an array
Import numpy as np and make an array `scores` of 88, 92, 70, 61, 95.

```python
# hint: import numpy as np , then np.array([...])
import numpy as np

scores = np.array([88, 92, 70, 61, 95])
print(scores)
print(type(scores))
```

### 2. Vectorized math
Add 5 bonus points to every score at once and print the result.

```python
# hint: scores + 5  (no loop needed)
print(scores + 5)
```

### 3. Summarize
Print the mean, the highest, and the lowest score.

```python
# hint: .mean() , .max() , .min()
print(scores.mean())
print(scores.max())
print(scores.min())
```

### 4. Boolean filter
Print only the scores above 80.

```python
# hint: scores[scores > 80]
print(scores[scores > 80])
```

### 5. Count (the challenge)
Print how many scores are above 80.

```python
# hint: (scores > 80).sum()
print((scores > 80).sum())
```
