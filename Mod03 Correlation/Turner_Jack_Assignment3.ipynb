3. Calculate and display Pearson's **correlation coefficients** and the **p-values** for *income* and *education*

```python
r, p = stats.pearsonr(df['income'], df['education'])
print(f"Correlation: {r:.3f}")
print(f"P-Value: {p:.3f}")
# Correlation:
# close to 1 or -1 = strong 
# close to 0 = weak

# P-value:
# anything less than 0.05 is statistically significant so it can be trusted.
# anything more is not significant enough to be trusted - and is likely unrelated or not related enough to matter.
```

6. Pearson's correlation coefficient OR covariance can both be used to measure the direction of the relationship between two variables (e.g., education and income). Explain why Pearson's correlation coefficient may be better than covariance.

Pearson's correlation coefficient is often better than covariance because it standardizes the relationship between two variables. Covariance tells us the direction of the relationship and how the variables change together, but its value depends on the units of measurement. For example, if income were measured in dollars instead of thousands of dollars, the covariance would change even though the underlying relationship stayed the same.

Pearson's correlation coefficient fixes this problem by dividing covariance by the product of the standard deviations of the two variables. This produces a value between -1 and 1, making it easier to compare relationships across different variables and scales. It shows both the direction and the strength of the linear relationship in a standardized way.

So, covariance tells whether the variables move together, but Pearson's correlation tells us how strong and in what direction the relationship is, regardless of units.
